```python
import random

# Fitness function
def fitness(x): return x*x

# Convert binary to decimal
def decode(chrom): return int("".join(map(str, chrom)), 2)

# Initialize population
pop = [[random.randint(0,1) for _ in range(5)] for _ in range(6)]

for gen in range(10):
    # Evaluate fitness
    fit = [fitness(decode(c)) for c in pop]

    # Selection (top 2)
    parents = [pop[i] for i in sorted(range(len(fit)), key=lambda i: fit[i], reverse=True)[:2]]

    # Crossover + Mutation
    new_pop = parents[:]
    while len(new_pop) < 6:
        p1, p2 = random.choice(parents), random.choice(parents)
        point = random.randint(1,4)
        child = p1[:point] + p2[point:]

        # Mutation
        if random.random() < 0.1:
            idx = random.randint(0,4)
            child[idx] ^= 1

        new_pop.append(child)

    pop = new_pop

# Best solution
best = max(pop, key=lambda c: fitness(decode(c)))
print("Best x:", decode(best), "Fitness:", fitness(decode(best)))
```
