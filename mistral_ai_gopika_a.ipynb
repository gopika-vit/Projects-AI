```python
!pip install -q mistralai
from mistralai import Mistral
import time

API_KEY = "SWTFAtjZ8vLvX36TIxetPpvXl15Mikt3"
MODEL_NAME = "mistral-tiny-latest"

# Create client
client = Mistral(api_key=API_KEY)


# -----------------------------------------------------
# SAFE API CALL for reproducibility (with retry)
# -----------------------------------------------------
def safe_api_call(prompt, seed=None, max_retries=5):
    for attempt in range(max_retries):
        try:
            response = client.chat.complete(
                model=MODEL_NAME,
                messages=[{"role": "user", "content": prompt}],
                temperature=0.0,
                random_seed=seed
            )
            return response.choices[0].message.content

        except Exception as e:
            msg = str(e)

            if "429" in msg or "capacity" in msg.lower():
                wait = 1 * (2 ** attempt)
                print(f"⚠ 429 Capacity issue → retry in {wait}s...")
                time.sleep(wait)
                continue

            raise e

    raise RuntimeError("❌ Failed all retries due to capacity overload.")


# -----------------------------------------------------
# NORMAL CHAT (no retry)
# -----------------------------------------------------
def chat_once(prompt):
    response = client.chat.complete(
        model=MODEL_NAME,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.0
    )
    return response.choices[0].message.content


# -----------------------------------------------------
# REPRODUCIBILITY TEST
# -----------------------------------------------------
def test_reproducibility(prompt, num_runs=3, seed=12345):
    print("\n===== AURA – REPRODUCIBILITY TEST =====")
    print("Model:", MODEL_NAME)
    print("Runs :", num_runs)
    print("Seed :", seed)
    print("Prompt:", prompt)
    print("=======================================\n")

    outputs = []

    for i in range(num_runs):
        print(f"▶ Run {i+1}...")
        text = safe_api_call(prompt, seed=seed)
        outputs.append(text)
        print(text, "\n")
        time.sleep(0.2)

    print("=======================================")
    if len(set(outputs)) == 1:
        print("✅ PERFECT — All outputs identical.")
    else:
        print("❌ NOT REPRODUCIBLE — Outputs differ.")
    print("=======================================")

    return outputs



# -----------------------------------------------------
# USER INPUT SECTION
# -----------------------------------------------------

# For reproducibility test
test_prompt = input("\nEnter a prompt for reproducibility testing → ")
test_reproducibility(test_prompt)
```
