```python
from deoldify import device
from deoldify.device_id import DeviceId
from deoldify.visualize import get_image_colorizer
```

```python
device.set(device=DeviceId.GPU0)
colorizer = get_image_colorizer(artistic=True)
```

```python
import os

INTERMEDIATE = "/content/drive/MyDrive/rop/restored/intermediate"
OUTPUT = "/content/drive/MyDrive/rop/restored/colorized"

os.makedirs(OUTPUT, exist_ok=True)
```

```python
from tqdm import tqdm

files = [f for f in os.listdir(INTERMEDIATE) if f.lower().endswith((".png",".jpg",".jpeg"))]

for fname in tqdm(files):
    path = os.path.join(INTERMEDIATE, fname)

    # Colorize
    result_path = colorizer.get_transformed_image(
        source=path,
        render_factor=35,
        watermarked=False,
        results_dir=OUTPUT
    )

print("All images colorized and saved to:", OUTPUT)
```
